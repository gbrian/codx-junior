## 2. Basic Imports & Setup

```python
import logging
from pprint import pprint
from ddgs import DDGS

# Set up logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s'
)
logger = logging.getLogger(__name__)

print("✅ DDGS library imported successfully")
print(f"DDGS version info: {DDGS.__doc__}")
```

## 3. Basic Text Search

```python
# Initialize DDGS instance
ddgs = DDGS()

# Perform a basic text search
query = "python programming"
print(f"\n🔍 Searching for: '{query}'\n")

results = ddgs.text(query, max_results=5)

print(f"📊 Found {len(results)} results:\n")
for idx, result in enumerate(results, 1):
    print(f"[{idx}] {result['title']}")
    print(f"    URL: {result['href']}")
    print(f"    Body: {result['body'][:150]}..." if len(result['body']) > 150 else f"    Body: {result['body']}")
    print()
```

## 8. Performance Comparison: DDGS vs Raw API

```python
import time
import requests

print("\n" + "="*70)
print("  PERFORMANCE COMPARISON: DDGS vs Raw API")
print("="*70 + "\n")

query = "machine learning"
num_runs = 3

# Test DDGS performance
print(f"🔄 Running {num_runs} iterations with DDGS...")
ddgs_times = []

for i in range(num_runs):
    start = time.time()
    ddgs_instance = DDGS()
    results = ddgs_instance.text(query, max_results=5)
    elapsed = time.time() - start
    ddgs_times.append(elapsed)
    print(f"  Run {i+1}: {elapsed:.3f}s (got {len(results)} results)")

print(f"\n  Average DDGS time: {sum(ddgs_times)/len(ddgs_times):.3f}s")

# Test Raw API performance
print(f"\n🔄 Running {num_runs} iterations with Raw API...")
raw_times = []

for i in range(num_runs):
    start = time.time()
    url = "https://api.duckduckgo.com/"
    params = {
        "q": query,
        "format": "json",
        "no_html": 1,
        "skip_disambig": 1,
    }
    response = requests.get(url, params=params, timeout=10)
    data = response.json()
    elapsed = time.time() - start
    raw_times.append(elapsed)
    print(f"  Run {i+1}: {elapsed:.3f}s (got {len(data.get('RelatedTopics', []))} topics)")

print(f"\n  Average Raw API time: {sum(raw_times)/len(raw_times):.3f}s")

print(f"\n📊 Summary:")
print(f"  DDGS average: {sum(ddgs_times)/len(ddgs_times):.3f}s")
print(f"  Raw API average: {sum(raw_times)/len(raw_times):.3f}s")
print(f"  Difference: {abs(sum(ddgs_times)/len(ddgs_times) - sum(raw_times)/len(raw_times)):.3f}s")
```

## 11. Integration with Search Tool

```python
# Proposed tool function using DDGS library

def duckduckgo_search_ddgs(query: str, max_results: int = 5) -> dict:
    """Perform a web search using DDGS library.
    
    Args:
        query: The search query.
        max_results: Maximum number of results to return.
    
    Returns:
        A dictionary containing results and metadata.
    """
    response = {
        'success': False,
        'query': query,
        'results': [],
        'error': None,
        'metadata': {
            'total_results': 0,
            'max_results': max_results,
        }
    }
    
    try:
        ddgs = DDGS(timeout=10)
        results = ddgs.text(query, max_results=max_results)
        
        # Transform results to consistent format
        for result in results:
            response['results'].append({
                'title': result.get('title', ''),
                'url': result.get('href', ''),
                'snippet': result.get('body', ''),
                'source': result.get('source', 'DuckDuckGo'),
            })
        
        response['success'] = True
        response['metadata']['total_results'] = len(response['results'])
    
    except Exception as e:
        response['error'] = f"{type(e).__name__}: {str(e)}"
        logger.error(f"DDGS search error: {response['error']}")
    
    return response

# Test the proposed tool function
print("\n" + "="*70)
print("  PROPOSED TOOL FUNCTION TEST")
print("="*70 + "\n")

test_result = duckduckgo_search_ddgs("python programming", max_results=3)
pprint(test_result)
```

## 12. Comparison: DDGS vs Raw API Results

```python
print("\n" + "="*70)
print("  COMPARISON: DDGS vs Raw API RESULTS")
print("="*70 + "\n")

query = "machine learning"

# Get DDGS results
print(f"🔍 Query: '{query}'\n")
print("DDGS Results:")
print("-" * 70)
ddgs_results = DDGS().text(query, max_results=3)
for idx, result in enumerate(ddgs_results, 1):
    print(f"[{idx}] {result['title']}")
    print(f"    URL: {result['href']}")
    print(f"    Snippet: {result['body'][:100]}...\n")

# Get Raw API results
print("\nRaw API Results:")
print("-" * 70)
url = "https://api.duckduckgo.com/"
params = {
    "q": query,
    "format": "json",
    "no_html": 1,
    "skip_disambig": 1,
}
response = requests.get(url, params=params, timeout=10)
data = response.json()

abstract = data.get('AbstractText', '')
if abstract:
    print(f"Abstract: {abstract[:100]}...\n")

related_topics = data.get('RelatedTopics', [])
for idx, topic in enumerate(related_topics[:3], 1):
    print(f"[{idx}] {topic['Text']}")
    print(f"    URL: {topic.get('FirstURL', 'N/A')}\n")

print("\n📊 OBSERVATIONS:")
print(f"  - DDGS provides more structured results with titles and snippets")
print(f"  - Raw API provides AbstractText and Related Topics")
print(f"  - DDGS results are more directly usable for web search")
print(f"  - Raw API results are more for entity/topic information")
```
